Lemmatization ---> more computational usage 
(Wordnet lemmatizer )

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
words=["going","absolutely","better","running","happier","faster"]
lemmatizer=WordNetLemmatizer()
for word in words:
    print(f"{word}--> {lemmatizer.lemmatize(word)}")
```

wordnetlemmatizer use parts_of_speech where if it is identifed as any "pos" it gives accurate/ exact root word after lemmatizing.
POS---> noun: 'n'
        verb:'v'
        adverb:'r'
        adjective:'a'

```python
lemmatizer.lemmatize("better", pos="a") ,lemmatizer.lemmatize("running", pos="v") 
```
