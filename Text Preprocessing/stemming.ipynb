Porter Stemmer

```python
import nltk
```

```python
from nltk.stem import PorterStemmer
porter_stemmer= PorterStemmer()
words=["successful","successfully","successes","successors","successfulness"]
for word in words:
    print(f"{word}---->{porter_stemmer.stem(word)}")
```

REGEXPSTEMMER CLASS---> use pattern to stem 

```python
from nltk.stem import RegexpStemmer
regex_stem=RegexpStemmer('ing$|es$|s$|ed$|ly$|ful$|ness$',min=3)
words=["successful","successfully","successes","successors","successfulness"]
for word in words:
    print(f"{word}---->{regex_stem.stem(word)}")
```

```python
regex_stem.stem("angrily")
```

SNOWBALLSTEMMER ---> better than porterstemmer

```python
from nltk.stem import SnowballStemmer 
snowball_stemmer=SnowballStemmer("english") #this stemmer is available for multiple languages
words=["successful","successfully","successes","successors","successfulness"]
```

```python
for word in words:
    print(f"{word}---->{snowball_stemmer.stem(word)}")
```
