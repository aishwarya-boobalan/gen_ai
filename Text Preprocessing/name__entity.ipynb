```python
text="""

On Monday, Aishwarya traveled from Chennai to Bengaluru to attend a technology conference organized by Microsoft. The event was held at the Bangalore International Exhibition Centre on August 15, 2026.

During the conference, she met Sundar Pichai, who spoke about the future of artificial intelligence and global technology. Several engineers from Google, Amazon, and OpenAI also attended the event.

Later, Aishwarya visited the Infosys campus and met her friend Priya Kumar. They discussed a new software project that could receive an investment of $5 million from a technology company based in California.

After the meeting, they traveled to Mysuru and visited the Mysore Palace. The next day, they returned to Chennai by train.

In September 2026, Aishwarya plans to visit London, United Kingdom, where she hopes to attend another international technology event. She also plans to meet researchers from Oxford University and visit several famous historical locations.
"""
```

```python
import nltk
```

```python
from nltk.tokenize import word_tokenize
words=word_tokenize(text)
```

```python
elements= nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
name_entities = nltk.ne_chunk(elements)
```
