```python
import gensim
```

```python
from gensim.models import Word2Vec,KeyedVectors
```

```python
import gensim.downloader as api
wv = api.load('word2vec-google-news-300')
```

```python
wv['happy']  # Get vector for word
```

```python
wv['cricket']  # Get vector for word
```

```python
wv.most_similar('king')
```

```python
wv.similarity('king', 'queen')  # Get similarity between two words
```

```python
vec=wv['king'] - wv['man'] + wv['woman']
```

```python
wv.most_similar([vec])
```
