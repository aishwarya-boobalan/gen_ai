```python
corpus="""This is my first sentence. \n
This is my second sentence. 
This is my third sentence. 
This is my fourth sentence. 
This is my fifth sentence. This is my sixth sentence. 
This is my seventh sentence. This is my eighth sentence. 
This is my ninth sentence. This is my tenth sentence.
"""
print(corpus)
```

```python
import nltk
nltk.download("punkt_tab")

```

```python
#sentence tokenization
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

```python
document=sent_tokenize(corpus)
for sent in document:
    print(sent)
    
#each sent is an element in the list where document is a list,and each element is a sentence in the list.
```

```python
from nltk.tokenize import  word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sent in document:
    print(word_tokenize(sent))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
